```python
import numpy as np
import matplotlib.pyplot as plt

N_data = np.loadtxt("n_data.txt")  # load the total number of syintetic data
h = np.loadtxt("mesh.txt")  # load the mesh sizes 
L2error_u = np.loadtxt("L2error_u.txt")  # load error in L2 norm for u
H1error_u = np.loadtxt("H1error_u.txt")  # load error in H1 norm for u
L2error_s = np.loadtxt("L2error_s.txt")  # load error in L2 norm for s

# Create subplots
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 6))

# Plotting L2 error for u
ax1.set_title('Error in L2 norm for u')
for i in range(len(N_data)):
    ax1.loglog(h, L2error_u[i, :], '*-', label='N = %.1e' % N_data[i]**2)
ax1.set_xlabel(r'$\Delta_x = \Delta_y = h$', fontsize=14)
ax1.set_ylabel(r'$\|u-u_{h}\|_{L2}$', fontsize=14)
ax1.grid(which='both', linestyle='--', linewidth=0.5, alpha=0.7)
ax1.minorticks_on()
ax1.legend(loc='lower right', fontsize=12)

# Plotting H1 error for u
ax2.set_title('Error in H1 norm for u')
for i in range(len(N_data)):
    ax2.loglog(h, H1error_u[i, :], 's-', label='N = %.1e' % N_data[i]**2)
ax2.set_xlabel(r'$\Delta_x = \Delta_y = h$', fontsize=14)
ax2.set_ylabel(r'$\|u-u_{h}\|_{H1}$', fontsize=14)
ax2.grid(which='both', linestyle='--', linewidth=0.5, alpha=0.7)
ax2.minorticks_on()
ax2.legend(loc='lower right', fontsize=12)

# Plotting L2 error for s
ax3.set_title('Error in L2 norm for p')
for i in range(len(N_data)):
    ax3.loglog(h, L2error_s[i, :], 'o-', label='N = %.1e' % N_data[i]**2)
ax3.set_xlabel(r'$\Delta_x = \Delta_y = h$', fontsize=14)
ax3.set_ylabel(r'$\|p-p_{h}\|_{L2}$', fontsize=14)
ax3.grid(which='both', linestyle='--', linewidth=0.5, alpha=0.7)
ax3.minorticks_on()
ax3.legend(loc='lower right', fontsize=12)

plt.tight_layout()
plt.show()
```
